```python
# Stacking Array
```

```python
a=np.array([1,2,3,4])
b=np.array([3,4,5,6])

# vertical stack 
c= np.vstack((a,b))   #input should give in one positional argument that is touple 
print(c)
```

```python
# horizontal stack in one single array other array also appends and resultant is ssingle array 
c= np.hstack((a,b))
print(c)
```

```python
# column stack 
c= np.column_stack((a,b))  # in matrix all the array act as a column 
print(c)
```
