```python
import numpy as np
import copy
```

```python
# Arithmatic operation 
```

```python
arr1= np.array([1,2,3,4,5])
arr2= np.array([6,7,8,9,10])

arr= arr1+arr2
print(arr)

# element number should be same 
```

```python
arr1-arr2
```

```python
arr1*arr2
```

```python
arr1//arr2
```

```python
# Broadcasting single value har elment pe add ya koi operation karna chhate ho toh broadcasting ka jaroorat padegi 
```

```python
l1= [1,2,4,5]
a1= np.array(l1)
print(a1)
a2= a1+10
print(a2)
```

```python
# similar broadcasting and arithmatic operation is done in matrix as well 
```

```python
a=np.array([[1,2],[3,4],[5,6]])
b= np.array([[7,8],[9,10],[11,12]])
c= a*b
print(c)
```

```python
arr1=np.array([1,2,3,4,5,6])
slice= arr1[::-1]
print(slice)
slice[0]=20
print(slice)
print(arr1)
```

```python
arr1=np.array([1,2,3,4,5,6])
slice= arr1[::-1]
print(slice)
slice= slice*10
print(slice)
print(arr1)
```

```python
arr1=np.array([1,2,3])
arr2=copy.deepcopy(arr1)
arr2[0]=12
print(arr1)
print(arr2)
```
