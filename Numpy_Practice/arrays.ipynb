```python
# Array methods 
```

```python
print(arr)
```

```python
print(arr.min())
```

```python
print(arr.max())
```

```python
print(arr.sum())
```

```python
# find the sum of array along vertical and horizontal 
sumofcolumns= np.sum(arr,axis=0)
print(sumofcolumns)
# axis =0 gives the sum of all the columns 

```

```python
sumofrows= np.sum(arr,axis=1)
print(sumofrows)
# gives the sum of all the rows 
```

```python
print(arr.mean())
```

```python
print(arr.std())
```

```python
print(arr.argmax())
# print the index of the max element 
```

```python
print(arr.argmin())
# print the index of the min element in the array
```
