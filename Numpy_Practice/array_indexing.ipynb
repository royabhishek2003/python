```python
# Numpy indexing and slicing og 2d matrix 
arr= np.arange(1,31).reshape(6,5);
print(arr);
print("\n");
print(arr[0,0]); #accessing the element in the 
print(arr[5,4]);

# slicing in one row 
# arr1= arr[0][0:4:1];
# arr2= arr[5][::-1];
# print(arr1);
# print(arr2);

# slicing for multiple rows 
arr3= arr[0:2, 1:3];
print(arr3);

arr4=arr[1:5, 1:4];  #remove all the border of a matrix 
print(arr4);

arr5= arr[:,2:3]; #here we are saying all rows and slice colum 3 to 3 means 3 so we get a matrix 
print(arr5)

arr5= arr[:,2]; # But here we are saying all rows and column 2 so here we get a oned array 
print(arr5)

```
